```python
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import normalize
from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer
from sklearn.metrics import silhouette_score
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import umap
import os


"""   

Copyright (c) 2026, Michael Tchuindjang
All rights reserved.

This code was developed as part of a PhD research project in Cybersecurity and Artificial Intelligence, 
supported by a studentship at the University of the West of England (UWE Bristol).

Use of this software is permitted for academic, educational, and research purposes.  
For any commercial use or redistribution, please contact the author for permission.

Disclaimer:
In no event shall the author or UWE be liable for any claim, damages, or other liability arising from the use of this code.

Acknowledgment of the author and the research context is appreciated in any derivative work or publication.


"""


# =========================
# GLOBAL PARAMETERS
# =========================
TRAIN_DIR = "training"
os.makedirs(TRAIN_DIR, exist_ok=True)

TESTS = ['Test_1', 
         'Test_2', 
         'Test_3', 
         'Test_4']

TRAINING_TEST = TESTS[1] # TESTS[1]: Test_2
# Tests/training files are formatted like Test_X_all_models.csv
TRAINING_TEST_FILE = TRAINING_TEST + '_all_models.csv'
INPUT_FILE = os.path.join(TRAIN_DIR, TRAINING_TEST_FILE)

# We automatically create the folder for ablation if missed
FOLDER = "ablation"
os.makedirs(FOLDER, exist_ok=True)

# =========================
# TEXT EXTRACTION
# =========================
def get_final_response(row):
    return row.get(f"output_turn_{row['turn_depth']}", "")

def get_full_conversation(row):
    conversation = []
    for i in range(1, row["turn_depth"] + 1):
        user_text = row.get(f"turn_{i}", "")
        assistant_text = row.get(f"output_turn_{i}", "")

        if pd.notna(user_text) and user_text.strip():
            conversation.append(f"USER: {user_text}")

        if pd.notna(assistant_text) and assistant_text.strip():
            conversation.append(f"ASSISTANT: {assistant_text}")

    return "\n".join(conversation)

# =========================
# SILHOUETTE SCORE
# =========================
def compute_silhouette(embeddings, n_clusters=10, random_state=42):
    # Safety check

    n = len(embeddings)
    n_clusters = min(n_clusters, max(2, int(np.sqrt(n))))
    
    if len(embeddings) < n_clusters:
        return -1

    labels = KMeans(
        n_clusters=n_clusters,
        random_state=random_state,
        n_init="auto"
    ).fit_predict(embeddings)

    # Must have at least 2 clusters
    if len(np.unique(labels)) < 2:
        return -1

    return silhouette_score(embeddings, labels)


# =========================
# UMAP SCORE
# =========================
def compute_umap(embeddings, n_neighbors=15, min_dist=0.1):
    reducer = umap.UMAP(
        n_neighbors=n_neighbors,
        min_dist=min_dist,
        metric="cosine",
        random_state=42
    )
    return reducer.fit_transform(embeddings)
    
def plot_umap_ablation(model_name, results_store):
    """
    Creates UMAP grid:
    rows = pooling methods
    cols = overlap values
    """

    overlaps = [0, 20, 40, 60]
    poolings = ["mean", "weighted", "max", "norm_mean"]

    fig, axes = plt.subplots(
        len(poolings),
        len(overlaps),
        figsize=(14, 10),
        sharex=True,
        sharey=True
    )

    for i, pooling in enumerate(poolings):
        for j, overlap in enumerate(overlaps):

            key = f"{model_name}_{pooling}_{overlap}"
            emb = results_store[key]

            emb_2d = compute_umap(emb)

            axes[i, j].scatter(
                emb_2d[:, 0],
                emb_2d[:, 1],
                s=8,
                alpha=0.7
            )

            if i == 0:
                axes[i, j].set_title(f"Overlap={overlap}")

            if j == 0:
                axes[i, j].set_ylabel(pooling)

            axes[i, j].set_xticks([])
            axes[i, j].set_yticks([])

    plt.suptitle(f"UMAP Ablation Grid - {model_name}", fontsize=14)
    plt.tight_layout()
    file_path = f"{FOLDER}/{TRAINING_TEST}_umap_{model_name}.png"
    plt.savefig(file_path, dpi=300, bbox_inches="tight")
    print(f"Saved: {file_path}")
    plt.show()

# =========================================================
# 1. LOAD MODELS
# =========================================================
models = {
    "MiniLM": {
        "model": SentenceTransformer("all-MiniLM-L6-v2"),
        "name": "sentence-transformers/all-MiniLM-L6-v2"
    },
    "MPNet": {
        "model": SentenceTransformer("all-mpnet-base-v2"),
        "name": "sentence-transformers/all-mpnet-base-v2"
    },
    "RoBERTa": {
        "model": SentenceTransformer("all-roberta-large-v1"),
        "name": "sentence-transformers/all-roberta-large-v1"
    }
}

# =========================================================
# 2. TOKEN-BASED CHUNKING (MODEL-CONSISTENT)
# =========================================================
def chunk_text_token_level(text, tokenizer, max_tokens=256, overlap=30):
    """
    Token-consistent chunking using model tokenizer.
    """
    tokens = tokenizer.encode(text, add_special_tokens=False)

    chunks = []
    start = 0

    while start < len(tokens):
        end = min(start + max_tokens, len(tokens))
        chunk_tokens = tokens[start:end]

        chunk = tokenizer.decode(chunk_tokens, skip_special_tokens=True)
        chunks.append(chunk.strip())

        start += max_tokens - overlap

    return chunks

# =========================================================
# 3. POOLING STRATEGIES (UNBIASED OPTIONS)
# =========================================================

def mean_pool(embeddings):
    return np.mean(embeddings, axis=0)

def weighted_mean_pool(embeddings, chunks, tokenizer):
    weights = np.array([
        len(tokenizer.encode(c, add_special_tokens=False))
        for c in chunks
    ])
    return np.average(embeddings, axis=0, weights=weights)

def max_pool(embeddings):
    return np.max(embeddings, axis=0)

def normalized_mean_pool(embeddings):
    emb = normalize(embeddings, axis=1)
    pooled = np.mean(emb, axis=0)
    return normalize(pooled.reshape(1, -1))[0]

# =========================================================
# 4. ENCODING FUNCTIONS
# =========================================================

def encode_no_chunk(texts, model):
    return model.encode(texts, convert_to_numpy=True, show_progress_bar=True)

def encode_chunk(texts, model, tokenizer, pooling="weighted", overlap=40, batch_size=64):
    all_embeddings = []
    chunk_counts = []
    all_chunks = []

    # -----------------------------
    # Flatten chunks
    # -----------------------------
    for text in texts:
        chunks = chunk_text_token_level(text, tokenizer, overlap=overlap)

        if len(chunks) == 0:
            chunks = [""]

        chunk_counts.append(len(chunks))
        all_chunks.extend(chunks)

    # -----------------------------
    # Encode chunks
    # -----------------------------
    chunk_embeddings = model.encode(
        all_chunks,
        batch_size=batch_size,
        convert_to_numpy=True,
        show_progress_bar=True
    )

    # -----------------------------
    # Pool per document
    # -----------------------------
    idx = 0

    for i, count in enumerate(chunk_counts):
        doc_chunks = chunk_embeddings[idx:idx + count]
        idx += count

        if pooling == "mean":
            doc_emb = mean_pool(doc_chunks)

        elif pooling == "weighted":
            doc_emb = weighted_mean_pool(doc_chunks, all_chunks[:count], tokenizer)

        elif pooling == "max":
            doc_emb = max_pool(doc_chunks)

        elif pooling == "norm_mean":
            doc_emb = normalized_mean_pool(doc_chunks)

        else:
            raise ValueError("Unknown pooling method")

        all_embeddings.append(doc_emb)

    return np.array(all_embeddings)

# =========================================================
# 5. LOAD DATA
# =========================================================
print(f"Loading data from {INPUT_FILE}...")
df = pd.read_csv(INPUT_FILE)

df["refusal"] = df.apply(get_final_response, axis=1)
texts = df["refusal"].dropna().tolist()
#texts = df["refusal"].dropna().tolist()[:10]

print(f"Loaded texts: {len(texts)}")

# =========================================================
# 6. EXPERIMENT LOOP
# =========================================================
results = []

OVERLAPS = [0, 20, 40, 60]
POOLING_METHODS = ["mean", "weighted", "max", "norm_mean"]

results = []
results_store = {}  # for UMAP visualization later

for name, model_info in models.items():
    print(f"\n==============================")
    print(f"Processing model: {name}")
    print(f"==============================")

    model = model_info["model"]
    tokenizer = AutoTokenizer.from_pretrained(model_info["name"])

    # -----------------------------
    # NO-CHUNK BASELINE
    # -----------------------------
    print("Encoding NO CHUNK...")
    no_chunk_embs = encode_no_chunk(texts, model)

    baseline_sil = compute_silhouette(no_chunk_embs, n_clusters=10)

    lengths = np.array([len(t.split()) for t in texts])

    results_store[f"{name}_no_chunk"] = no_chunk_embs

    print("Encoding CHUNKED...")
    # -----------------------------
    # CHUNKED ABLATION
    # -----------------------------
    for overlap in OVERLAPS:
        print(f"\n--- Overlap: {overlap} ---")

        for pooling in POOLING_METHODS:
            print(f"Pooling: {pooling}")

            # -------------------------
            # Encode chunked texts
            # -------------------------
            chunk_embs = encode_chunk(
                texts,
                model,
                tokenizer,
                pooling=pooling,
                overlap=overlap
            )

            # store for UMAP
            key = f"{name}_{pooling}_{overlap}"
            results_store[key] = chunk_embs

            # -------------------------
            # Similarity vs baseline
            # -------------------------
            sims = cosine_similarity(no_chunk_embs, chunk_embs).diagonal()

            # -------------------------
            # Silhouette score (structure quality)
            # -------------------------
            sil = compute_silhouette(chunk_embs, n_clusters=10)

            # -------------------------
            # Metrics
            # -------------------------
            results.append({
                "model": name,
                "overlap": overlap,
                "pooling": pooling,
                "mean_similarity": np.mean(sims),
                "std_similarity": np.std(sims),
                "length_corr": np.corrcoef(lengths, sims)[0, 1],
                "silhouette": sil,
                "baseline_silhouette": baseline_sil
            })

# =========================================================
# 7. RESULTS
# =========================================================
results_df = pd.DataFrame(results)
print("\n===== TOKEN-CONSISTENT CHUNKING POOLING ABLATION RESULTS =====")
print(results_df)

results_df["score"] = (
    0.3 * results_df["mean_similarity"]
    + 0.2 * (1 - results_df["std_similarity"])
    + 0.2 * (1 - results_df["length_corr"].abs())
    + 0.3 * results_df["silhouette"]
)

best_balanced = results_df.loc[results_df["score"].idxmax()]

print("\n===== BEST BALANCED CONFIGURATION =====")
print(best_balanced)
results_df.to_csv(f"{FOLDER}/{TRAINING_TEST}_chunking_pooling_ablation_results.csv", index=False)
print(f"\nSaved to: {FOLDER}/{TRAINING_TEST}_chunking_pooling_ablation_results.csv")

# =========================
# UMAP VISUALISATION
# =========================
plot_umap_ablation("MiniLM", results_store)
plot_umap_ablation("MPNet", results_store)
plot_umap_ablation("RoBERTa", results_store)
```
